```python
import os
import sys
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import geopandas as gpd
```

```python
#load bc municipal boundaries shapefile
gdf_municipal_boundaries = gpd.read_file('../data/raw/BC_municipal_boundaries/158_BC_municipal_boundaries.shp')
#load final analysis data
df_final = pd.read_csv('../data/processed/final_analysis_data_with_coords.csv')
```

```python
#see which municipalities are in the final data but not in the shapefile
missing_municipalities = set(df_final['Community_Name'].unique()) - set(gdf_municipal_boundaries['ABRVN'].unique())
if missing_municipalities:
    print("Municipalities in the final data but not in the shapefile:")
    print(missing_municipalities)
```

```python
#print all communities in the shapefile
all_communities = set(gdf_municipal_boundaries['ABRVN'].unique())
print("All communities in the shapefile:")
for community in sorted(all_communities):
    print(community)
```
